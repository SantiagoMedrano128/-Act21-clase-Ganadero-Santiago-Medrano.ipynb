#### el programa genere automáticamente opiniones o "consejos de IA" basados en los atributos del cantante

```python
class Cantante:
    def __init__(self, nombre, registro_vocal, albumes, genero, giras, sello, fans):
        self.nombre = nombre
        self.registro_vocal = registro_vocal
        self.albumes = int(albumes)
        self.genero = genero
        self.giras = int(giras)
        self.sello = sello
        self.fans = int(fans)

    def simular_ia(self):
        # Lógica de simulación de IA basada en las estadísticas del cantante
        if self.fans > 1000000 and self.giras > 5:
            return "🤖 IA: ¡Superestrella Global! Recomendación: Lanzar mercancía premium y gira de estadios."
        elif self.albumes > 3 and self.fans < 100000:
            return "🤖 IA: Joya Oculta del Indie. Recomendación: Invertir fuertemente en campañas de TikTok."
        else:
            return "🤖 IA: Talento en Desarrollo. Recomendación: Lanzar colaboraciones (Feat.) para ganar tracción."


class DisqueraIA:
    def __init__(self):
        # Iniciamos la lista completamente vacía a petición
        self.lista_cantantes = []

    def agregar_cantante(self):
        print("\n" + "═"*40)
        print("🎤   REGISTRAR NUEVO TALENTO   🎤")
        print("═"*40)
        nombre = input("▶ Nombre del cantante: ")
        registro = input("▶ Registro vocal (ej. Tenor, Soprano): ")
        albumes = input("▶ Número de álbumes grabados: ")
        genero = input("▶ Género musical: ")
        giras = input("▶ Número de giras realizadas: ")
        sello = input("▶ Sello discográfico: ")
        fans = input("▶ Número de fans (seguidores): ")

        nuevo = Cantante(nombre, registro, albumes, genero, giras, sello, fans)
        self.lista_cantantes.append(nuevo)
        print(f"\n✅ ¡{nombre} ha sido firmado exitosamente en el sistema!")

    def listar_cantantes(self):
        # Si la lista está vacía, mostramos un mensaje amigable en lugar de una tabla en blanco
        if not self.lista_cantantes:
            print("\n" + "─"*55)
            print("📭 La base de datos está vacía. ¡Registra un cantante primero!")
            print("─"*55)
            return

        # Diseño de la tabla perfectamente alineada y distribuida
        # Los números como <16 indican el ancho en caracteres de cada columna
        print("\n" + "╔" + "═"*119 + "╗")
        print(f"║ {'CANTANTE':<16} │ {'REG. VOCAL':<12} │ {'ÁLBUMES':<8} │ {'GÉNERO':<16} │ {'GIRAS':<7} │ {'SELLO':<15} │ {'FANS':<15} ║")
        print("╠" + "═"*119 + "╣")

        for c in self.lista_cantantes:
            # {c.fans:,} añade automáticamente comas para separar los miles en los números (ej: 1,500,000)
            print(f"║ {c.nombre:<16} │ {c.registro_vocal:<12} │ {c.albumes:<8} │ {c.genero:<16} │ {c.giras:<7} │ {c.sello:<15} │ {c.fans:<15,} ║")

        print("╚" + "═"*119 + "╝")

    def actualizar_cantante(self):
        if not self.lista_cantantes:
            print("\n❌ No hay cantantes registrados para actualizar.")
            return

        print("\n" + "═"*40)
        print("🔄   ACTUALIZAR DATOS DE ARTISTA   🔄")
        print("═"*40)
        nombre_buscar = input("Introduce el nombre del cantante a modificar: ")
        encontrado = False

        for c in self.lista_cantantes:
            if c.nombre.lower() == nombre_buscar.lower():
                encontrado = True
                print(f"\n🎯 ¡Artista encontrado! [Presiona ENTER para mantener el valor actual]")

                nuevo_registro = input(f"Nuevo registro vocal [{c.registro_vocal}]: ")
                if nuevo_registro: c.registro_vocal = nuevo_registro

                nuevo_albumes = input(f"Nuevos álbumes [{c.albumes}]: ")
                if nuevo_albumes: c.albumes = int(nuevo_albumes)

                nuevo_genero = input(f"Nuevo género [{c.genero}]: ")
                if nuevo_genero: c.genero = nuevo_genero

                nuevo_giras = input(f"Nuevas giras [{c.giras}]: ")
                if nuevo_giras: c.giras = int(nuevo_giras)

                nuevo_sello = input(f"Nuevo sello [{c.sello}]: ")
                if nuevo_sello: c.sello = nuevo_sello

                nuevos_fans = input(f"Nuevos fans [{c.fans}]: ")
                if nuevos_fans: c.fans = int(nuevos_fans)

                print(f"\n✨ ¡Los datos de {c.nombre} se han actualizado correctamente!")
                break

        if not encontrado:
            print("❌ No se encontró ningún cantante con ese nombre.")

    def borrar_cantante(self):
        if not self.lista_cantantes:
            print("\n❌ No hay cantantes en el sistema para eliminar.")
            return

        print("\n" + "═"*40)
        print("❌   ELIMINAR CANTANTE (RESCINDIR CONTRATO)   ❌")
        print("═"*40)
        nombre_buscar = input("Introduce el nombre del cantante a dar de baja: ")
        encontrado = False

        for c in self.lista_cantantes:
            if c.nombre.lower() == nombre_buscar.lower():
                self.lista_cantantes.remove(c)
                print(f"\n🔥 Contrato cancelado. {c.nombre} ha sido eliminado del sistema.")
                encontrado = True
                break

        if not encontrado:
            print("❌ No se encontró ningún cantante con ese nombre.")

    def consultar_ia_menu(self):
        if not self.lista_cantantes:
            print("\n❌ No hay cantantes en el sistema para analizar.")
            return

        print("\n" + "═"*40)
        print("🧠   ANALIZADOR DE TALENTO (IA)   🧠")
        print("═"*40)
        nombre_buscar = input("¿A qué cantante deseas que la IA analice?: ")
        encontrado = False

        for c in self.lista_cantantes:
            if c.nombre.lower() == nombre_buscar.lower():
                print(f"\n📊 Reporte de Inteligencia Artificial para: {c.nombre}")
                print(c.simular_ia())
                encontrado = True
                break
        if not encontrado:
            print("❌ Cantante no encontrado.")


# --- MENÚ INTERACTIVO PRINCIPAL ---
disquera = DisqueraIA()
continuar = True

while continuar:
    print("\n🎧 " + "═"*45 + " 🎤")
    print("      SISTEMA CYBER-MUSIC v2.0 (CON IA)      ")
    print("═"*51)
    print(" 1. ➕ Agregar Cantante")
    print(" 2. 📋 Listar Cantantes (Ver Tabla)")
    print(" 3. 🔄 Actualizar Cantante")
    print(" 4. ❌ Borrar Cantante")
    print(" 5. 🧠 Consultar a la IA Predictiva")
    print(" 6. 🚪 Salir")
    print("═"*51)

    opcion = input("Selecciona una opción (1-6): ")

    if opcion == "1":
        disquera.agregar_cantante()
    elif opcion == "2":
        disquera.listar_cantantes()
    elif opcion == "3":
        disquera.actualizar_cantante()
    elif opcion == "4":
        disquera.borrar_cantante()
    elif opcion == "5":
        disquera.consultar_ia_menu()
    elif opcion == "6":
        print("\n👋 ¡Gracias por usar Cyber-Music! Apagando sistemas...")
        continuar = False
    else:
        print("\n⚠ Opción no válida. Por favor, introduce un número del 1 al 6.")
```
